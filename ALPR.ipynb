```python
import sqlite3
import ipywidgets as widgets
from IPython.display import display

# Fonction pour vérifier si la plaque est autorisée
def verifier_acces(plaque):
    conn = sqlite3.connect('/content/drive/MyDrive/Licence-Plate-Detection-using-YOLO-V8/alpr.db')
    cursor = conn.cursor()
    cursor.execute("SELECT COUNT(*), nom_proprietaire FROM immatriculations WHERE immatriculation = ?", (plaque,))
    result = cursor.fetchone()
    conn.close()
    return result[0] > 0, result[1]

# Plaque d'immatriculation à vérifier
plaque_a_verifier = ocr_results[0]

# Affichage du message
def afficher_message(plaque):
    access, name = verifier_acces(plaque)

    if access:
        message = f"Accès autorisé! Bonjour {name}"
        message_widget.value = f'<div style="background-color: lightgreen; padding: 20px; border-radius: 5px; font-size: 16pt; font-weight: bold; color: green;">{message}</div>'
    else:
        message = "Accès refusé"
        message_widget.value = f'<div style="background-color: lightcoral; padding: 20px; border-radius: 5px; font-size: 16pt; font-weight: bold; color: red;">{message}</div>'

# Widget HTML pour afficher le message
message_widget = widgets.HTML(
    value="",
    placeholder='',
    description=''
)

# Afficher le message pour la plaque spécifiée
afficher_message(plaque_a_verifier)

# Afficher le widget
display(message_widget)
```
